```python
import pandas as pd

file_path = '../Data/bank/bank-full.csv'
df = pd.read_csv(file_path, sep=';')

df.head()
```

```python
print("Dataset Shape:", df.shape)
df.info()
```

```python

print("Class Counts:")
print(df['y'].value_counts())

print("\nClass Percentage (%):")
print(df['y'].value_counts(normalize=True) * 100)
```

```python
print("--- Numerical Features Summary ---")
display(df.describe())

print("\n--- Categorical Features Summary ---")
display(df.describe(include='object'))
```

```python
print("--- Null/NaN Values Count ---")
print(df.isnull().sum())

print("\n--- 'unknown' Values in Categorical Features ---")
for col in df.select_dtypes(include='object').columns:
    count = (df[col] == 'unknown').sum()
    if count > 0:
        percentage = (count / len(df)) * 100
        print(f"{col}: {count} rows ({percentage:.2f}%)")

no_prev_contact = (df['pdays'] == -1).sum()
print(f"\n--- pdays Sentinel Values (-1) ---")
print(f"pdays == -1: {no_prev_contact} rows ({(no_prev_contact/len(df))*100:.2f}%)")
```
